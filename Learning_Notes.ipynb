```python
from azure.identity import InteractiveBrowserCredential, AzureCliCredential
from azure.ai.ml import MLClient

# Pick ONE of these credential types:

# A) Interactive browser flow locked to your tenant
cred = InteractiveBrowserCredential(
    tenant_id="0c2d2521-cc09-4a90-98a7-ee569058250d"
)

# B) Or: reuse `az login` but still lock to tenant (works if your az context is correct)
# cred = AzureCliCredential(tenant_id="0c2d2521-cc09-4a90-98a7-ee569058250d")

ml_client = MLClient(
    credential=cred,
    subscription_id="4b55259d-2ceb-46a9-bf52-5a7712cd576c",
    resource_group_name="rg-dp100-labs",
    workspace_name="mlw-dp100-labs-2",
)
```

## Explore and configure the Azure Machine Learning workspace

```python
# list the environments using the Python SDK:
envs = ml_client.environments.list()
for env in envs:
    print(env.name)

# To review the details of a specific environment, you can retrieve an environment by its registered name:
env = ml_client.environments.get(name="AzureML-ACPT-pytorch-1.13-py38-cuda11.7-gpu", version="1")
print(env)
```
